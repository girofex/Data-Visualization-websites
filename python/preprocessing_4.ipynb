### Choropleth

```python
events = "Riots", "Explosions/Remote violence", "Violence against civilians"

def choropleth(df):
    df = df[df["EVENT_TYPE"].isin(events)]
    df = df.drop(columns=["WEEK", "REGION", "ADMIN1", "SUB_EVENT_TYPE", "FATALITIES", 
                          "POPULATION_EXPOSURE", "DISORDER_TYPE", "ID", 
                          "CENTROID_LATITUDE", "CENTROID_LONGITUDE"])
    
    event_counts = df.groupby(["COUNTRY", "EVENT_TYPE"]).size().reset_index(name='count')
    dominant_event = event_counts.loc[event_counts.groupby("COUNTRY")['count'].idxmax()]
    result = dominant_event[["COUNTRY", "EVENT_TYPE"]].copy()
    
    replacements = {
        'United States': 'USA',
        'United Kingdom': 'England',
        'Democratic Republic of Congo': 'Democratic Republic of the Congo',
        'Republic of Congo': 'Republic of the Congo',
        'Tanzania': 'United Republic of Tanzania',
        'eSwatini': 'Swaziland',
        'Guinea-Bissau': 'Guinea Bissau',
        'North Macedonia': 'Macedonia',
        'Serbia': 'Republic of Serbia',
        'Bahamas': 'The Bahamas',
        'Palestine': 'West Bank'
    }
    
    result["COUNTRY"] = result["COUNTRY"].replace(replacements)
    
    return result
```

```python
africa = pd.read_csv("../resources/og/Africa_aggregated_data_up_to-2025-10-18.csv")
asiapacific = pd.read_csv("../resources/og/Asia-Pacific_aggregated_data_up_to-2025-10-11_0.csv")
europecentralasia = pd.read_csv("../resources/og/Europe-Central-Asia_aggregated_data_up_to-2025-10-11.csv")
latinamericacaribbean = pd.read_csv("../resources/og/Latin-America-the-Caribbean_aggregated_data_up_to-2025-10-18.csv")
middleeast = pd.read_csv("../resources/og/Middle-East_aggregated_data_up_to-2025-10-18.csv")
uscanada = pd.read_csv("../resources/og/US-and-Canada_aggregated_data_up_to-2025-10-11_0.csv")
```

```python
africa = choropleth(africa)
asiapacific = choropleth(asiapacific)
europecentralasia = choropleth(europecentralasia)
latinamericacaribbean = choropleth(latinamericacaribbean)
middleeast = choropleth(middleeast)
uscanada = choropleth(uscanada)
```

```python
choropleth = pd.concat([africa, asiapacific, europecentralasia, latinamericacaribbean, middleeast, uscanada])
choropleth.to_csv("../resources/plots/sectionfour/choropleth.csv", index=False)
```

## Dot map

```python
countries = ["Russia", "Ukraine", "Israel", "Palestine", "Mexico"]

def dot(df):
    df = df[df["EVENT_TYPE"].isin(events)]
    df = df[df["COUNTRY"].isin(countries)]
    df = df.drop(columns=["REGION", "ADMIN1", "SUB_EVENT_TYPE", "FATALITIES", 
                          "POPULATION_EXPOSURE", "DISORDER_TYPE", "ID"])
    df_grouped = df.groupby(["COUNTRY", "CENTROID_LATITUDE", "CENTROID_LONGITUDE", "EVENT_TYPE"])["EVENTS"].sum() # added now
    df = df_grouped.reset_index()
    
    return df
```

```python
europecentralasia = pd.read_csv("../resources/og/Europe-Central-Asia_aggregated_data_up_to-2025-10-11.csv")
latinamericacaribbean = pd.read_csv("../resources/og/Latin-America-the-Caribbean_aggregated_data_up_to-2025-10-18.csv")
middleeast = pd.read_csv("../resources/og/Middle-East_aggregated_data_up_to-2025-10-18.csv")
```

```python
europecentralasia = dot(europecentralasia)
latinamericacaribbean = dot(latinamericacaribbean)
middleeast = dot(middleeast)
```

```python
europecentralasia.to_csv("../resources/plots/sectionfour/dot1_alt.csv", index=False)
latinamericacaribbean.to_csv("../resources/plots/sectionfour/dot2_alt.csv", index=False)
middleeast.to_csv("../resources/plots/sectionfour/dot3_alt.csv", index=False)
```

# Hexbin map

```python
def hex(df):
    df = df[df["EVENT_TYPE"].isin(events)]
    
    df = df.drop(columns=[
        "WEEK", "REGION", "ADMIN1", "SUB_EVENT_TYPE", "FATALITIES",
        "POPULATION_EXPOSURE", "DISORDER_TYPE", "ID",
        "CENTROID_LATITUDE", "CENTROID_LONGITUDE"
    ])
    
    event_counts = df.groupby("COUNTRY").size().reset_index(name="count")

    replacements = {
        'United States': 'USA',
        'United Kingdom': 'England',
        'Democratic Republic of Congo': 'Democratic Republic of the Congo',
        'Republic of Congo': 'Republic of the Congo',
        'Tanzania': 'United Republic of Tanzania',
        'eSwatini': 'Swaziland',
        'Guinea-Bissau': 'Guinea Bissau',
        'North Macedonia': 'Macedonia',
        'Serbia': 'Republic of Serbia',
        'Bahamas': 'The Bahamas',
        'Palestine': 'West Bank'
    }
    
    event_counts["COUNTRY"] = event_counts["COUNTRY"].replace(replacements)

    return event_counts
```

```python
africa = pd.read_csv("../resources/og/Africa_aggregated_data_up_to-2025-10-18.csv")
asiapacific = pd.read_csv("../resources/og/Asia-Pacific_aggregated_data_up_to-2025-10-11_0.csv")
europecentralasia = pd.read_csv("../resources/og/Europe-Central-Asia_aggregated_data_up_to-2025-10-11.csv")
latinamericacaribbean = pd.read_csv("../resources/og/Latin-America-the-Caribbean_aggregated_data_up_to-2025-10-18.csv")
middleeast = pd.read_csv("../resources/og/Middle-East_aggregated_data_up_to-2025-10-18.csv")
uscanada = pd.read_csv("../resources/og/US-and-Canada_aggregated_data_up_to-2025-10-11_0.csv")
```

```python
africa = hex(africa)
asiapacific = hex(asiapacific)
europecentralasia = hex(europecentralasia)
latinamericacaribbean = hex(latinamericacaribbean)
middleeast = hex(middleeast)
uscanada = hex(uscanada)
```

```python
hex = pd.concat([africa, asiapacific, europecentralasia, latinamericacaribbean, middleeast, uscanada])
hex.to_csv("../resources/plots/sectionfour/hex.csv", index=False)
```
