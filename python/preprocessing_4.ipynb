### Choropleth

```python
events = "Riots", "Explosions/Remote violence", "Violence against civilians"

def choropleth(df):
    df = df[df["EVENT_TYPE"].isin(events)]
    df = df.drop(columns=["WEEK", "REGION", "ADMIN1", "SUB_EVENT_TYPE", "FATALITIES", 
                          "POPULATION_EXPOSURE", "DISORDER_TYPE", "ID", 
                          "CENTROID_LATITUDE", "CENTROID_LONGITUDE"])
    
    event_counts = df.groupby(["COUNTRY", "EVENT_TYPE"]).size().reset_index(name='count')
    dominant_event = event_counts.loc[event_counts.groupby("COUNTRY")['count'].idxmax()]
    result = dominant_event[["COUNTRY", "EVENT_TYPE"]].copy()
    
    result['COUNTRY'] = result['COUNTRY'].replace('United States', 'USA')
    result['COUNTRY'] = result['COUNTRY'].replace('United Kingdom', 'England')
    result['COUNTRY'] = result['COUNTRY'].replace('Democratic Republic of Congo', 'Democratic Republic of the Congo')
    result['COUNTRY'] = result['COUNTRY'].replace('Republic of Congo', 'Republic of the Congo')
    result['COUNTRY'] = result['COUNTRY'].replace('Tanzania', 'United Republic of Tanzania')
    result['COUNTRY'] = result['COUNTRY'].replace('eSwatini', 'Swaziland')
    result['COUNTRY'] = result['COUNTRY'].replace('Guinea-Bissau', 'Guinea Bissau')
    result['COUNTRY'] = result['COUNTRY'].replace('North Macedonia', 'Macedonia')
    result['COUNTRY'] = result['COUNTRY'].replace('Serbia', 'Republic of Serbia')
    result['COUNTRY'] = result['COUNTRY'].replace('Bahamas', 'The Bahamas')
    
    return result
```

```python
africa = pd.read_csv("../resources/og/Africa_aggregated_data_up_to-2025-10-18.csv")
asiapacific = pd.read_csv("../resources/og/Asia-Pacific_aggregated_data_up_to-2025-10-11_0.csv")
europecentralasia = pd.read_csv("../resources/og/Europe-Central-Asia_aggregated_data_up_to-2025-10-11.csv")
latinamericacaribbean = pd.read_csv("../resources/og/Latin-America-the-Caribbean_aggregated_data_up_to-2025-10-18.csv")
middleeast = pd.read_csv("../resources/og/Middle-East_aggregated_data_up_to-2025-10-18.csv")
uscanada = pd.read_csv("../resources/og/US-and-Canada_aggregated_data_up_to-2025-10-11_0.csv")
```

```python
africa = choropleth(africa)
asiapacific = choropleth(asiapacific)
europecentralasia = choropleth(europecentralasia)
latinamericacaribbean = choropleth(latinamericacaribbean)
middleeast = choropleth(middleeast)
uscanada = choropleth(uscanada)
```

```python
choropleth = pd.concat([africa, asiapacific, europecentralasia, latinamericacaribbean, middleeast, uscanada])
choropleth.to_csv("../resources/plots/sectionfour/choropleth.csv", index=False)
```
